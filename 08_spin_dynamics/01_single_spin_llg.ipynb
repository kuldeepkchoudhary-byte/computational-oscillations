# Single Spin LLG

## Theory
Precession and damping.

```python
import numpy as np
import matplotlib.pyplot as plt
from coupled_oscillators.spin_dynamics import integrate_llg
from coupled_oscillators.plotting import plot_time_series

params = {"gamma": 1.0, "alpha": 0.1, "H_ext": np.array([0.0, 0.0, 1.0]), "J": 0.0, "K_ani": 0.0}
t_grid = np.linspace(0, 50, 500)
M0 = np.array([1.0, 0.0, 0.0])

history = integrate_llg(M0, t_grid, params)
fig, ax = plot_time_series(t_grid, history, labels=["Mx", "My", "Mz"], title="Single Spin LLG Damping")
import os
os.makedirs('01_single_spin_llg', exist_ok=True)
plt.savefig(f'01_single_spin_llg/plot_{plt.gcf().number}.png')
plt.show()
```
